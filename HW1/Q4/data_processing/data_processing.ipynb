# TAQ Data Processor Performance Comparison

Comparing processing times for Pandas vs Polars implementations on MRNA stock data.

```python
import time
import sys
from pathlib import Path

sys.path.append('../..')
from utils import TAQDataProcessorPandas, TAQDataProcessorPolars
```

## Test Configuration

```python
# Paths to MRNA data (relative to notebook location in Q4/data_processing/)
MRNA_TRADES = "../data/TAQ_MRNA_trades_sep2025.csv"
MRNA_QUOTES = "../data/TAQ_MRNA_quotes_sep2025.csv"

# Output paths
MRNA_TRADES_PANDAS_OUTPUT = "./results/MRNA_trades_pandas.parquet"
MRNA_TRADES_POLARS_OUTPUT = "./results/MRNA_trades_polars.parquet"
MRNA_QUOTES_PANDAS_OUTPUT = "./results/MRNA_quotes_pandas.parquet"
MRNA_QUOTES_POLARS_OUTPUT = "./results/MRNA_quotes_polars.parquet"

# Create results directory
Path("./results").mkdir(parents=True, exist_ok=True)

print("Configuration ready:")
print(f"  Trades input: {MRNA_TRADES}")
print(f"  Quotes input: {MRNA_QUOTES}")
print(f"  Output directory: ./results/")
```

## Trade Data Processing Comparison

```python
print("\n" + "="*70)
print("TESTING TRADE DATA PROCESSING")
print("="*70)

# Pandas implementation
print("\nPandas Processing:")
pandas_processor = TAQDataProcessorPandas()
start = time.time()
pandas_processor.process_trades(MRNA_TRADES, MRNA_TRADES_PANDAS_OUTPUT)
pandas_time = time.time() - start
print(f"  Time: {pandas_time:.2f} seconds")

# Polars implementation
print("\nPolars Processing:")
polars_processor = TAQDataProcessorPolars()
start = time.time()
polars_processor.process_trades(MRNA_TRADES, MRNA_TRADES_POLARS_OUTPUT)
polars_time = time.time() - start
print(f"  Time: {polars_time:.2f} seconds")

# Comparison
speedup = pandas_time / polars_time
print(f"\nComparison:")
print(f"  Speedup (Pandas time / Polars time): {speedup:.2f}x")
if speedup > 1:
    print(f"  ✓ Polars is {speedup:.2f}x faster")
else:
    print(f"  ✓ Pandas is {1/speedup:.2f}x faster")
```

## Quote Data Processing Comparison

```python
print("\n" + "="*70)
print("TESTING QUOTE DATA PROCESSING")
print("="*70)

# Pandas implementation
print("\nPandas Processing:")
pandas_processor = TAQDataProcessorPandas()
start = time.time()
pandas_processor.process_quotes(MRNA_QUOTES, MRNA_QUOTES_PANDAS_OUTPUT)
pandas_time = time.time() - start
print(f"  Time: {pandas_time:.2f} seconds")

# Polars implementation
print("\nPolars Processing:")
polars_processor = TAQDataProcessorPolars()
start = time.time()
polars_processor.process_quotes(MRNA_QUOTES, MRNA_QUOTES_POLARS_OUTPUT)
polars_time = time.time() - start
print(f"  Time: {polars_time:.2f} seconds")

# Comparison
speedup = pandas_time / polars_time
print(f"\nComparison:")
print(f"  Speedup (Pandas time / Polars time): {speedup:.2f}x")
if speedup > 1:
    print(f"  ✓ Polars is {speedup:.2f}x faster")
else:
    print(f"  ✓ Pandas is {1/speedup:.2f}x faster")
```
